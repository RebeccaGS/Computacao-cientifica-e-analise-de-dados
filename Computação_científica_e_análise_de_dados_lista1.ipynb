## E1
**(a) Gere aleatoriamente 30 pontos de um polinômio de grau 5**

Para isso, vou gerar 30 pares de pontos $(x_1,y_1),...,(x_n,y_n)$ no polinômio $p(x) = a_5x^5 + a_4x^4 + a_3x^3 + a_2x^2 + a_1x + a_0$
- Primeira coisa é criar um polinômio de grau 5, escolhi: $p(x) = 2x^5+4$;
- Depois, gerar 30 x's aleatórios, usando a função rand;
- e descobrir seus respectivos y's.



```julia
x=rand(big(1):100,30)
println(x)
y= evalpoly.(x,((4,0,0,0,0,2),)) #p(x) = 4 + 2x^5
println(y)
```

## E1
**(b) Faça regressão polinomial com polinômios de grau 0 até 29.**
- Tenho 30 pares de pontos, quero achar polinômios que se aproximam desses pontos.
- Um polinômio de grau 0: $a_0x^0$ (reta horizontal que tente se aproximar dos 30 pontos e corta y em $a_0$); Grau 1: $a_0x^0+a_1x^1$ (reta com inclinação $a_1$ que corte y em $a_0$); ...

Achar meu $p(x) =  \sum_0^n a_nx^n$ que fique mais próximo possível dos 30 pontos, com n representando o grau de 0 a 29.
- $y_i' = p(x_i)$, logo, $e_i = y_i - y_i'$
- **Método dos mínimos quadrados**: Procura coeficientes $a_0, ..., a_n$ que faz soma dos quadrados dos erros ser a menor possível $min \sum_1^{30}(y_i'-p(x_i))^2$
- Em grau2: $a_0x^0+a_1x^1+a_2x^2$ --> substituindo 30 x's, não consguiria satisfazer as 30 equações ao mesmo tempo --> usar mínimos quadrados para achar melhor ajuste
- $$\begin{bmatrix} 1 & x_1 & x_1^2 \\ 1 & x_2 & x_2^2 \\ \vdots & \vdots & \vdots \\ 1 & x_{30} & x_{30}^2 \end{bmatrix}
\begin{bmatrix} a_0 \\ a_1 \\ a_2 \end{bmatrix}
\approx
\begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_{30} \end{bmatrix} $$
- $Xa \approx y$
- Sempre 30 linhas (30 x's), muda número de colunas
- grau 0 → 1 coluna  → a₀
- grau 1 → 2 colunas → a₀, a₁
- grau 2 → 3 colunas → a₀, a₁, a₂
- grau 29 → 30 colunas

```julia
using LinearAlgebra
erros = []

for grau in 0:29
  matriz = [xi^n for xi in x, n in 0:grau] #linhas: xi in x //colunas: n in 0:grau//xi^n conteudo
  a = matriz \ y
  a_limpo = round.(Float64.(a), digits=4)
  #println(a_limpo)

  erro = norm(matriz*a-y)
  push!(erros, erro)
  println("Grau $grau - Erro: ", Float64(erro))
end
```

```julia
using LinearAlgebra

#Defino Toy Stpry como meu 0, faço a matriz sem T

    #V  R  C  I  D  B  H  S
A = [0 -1  0  0  0  0  0  0; #T-R=11 T=0
     1  0 -1  0  0  0  0  0; #V-C=3
     0  0  0  1 -1  0  0  0; #I-D=7
     0  0  0  0  0  1 -1  0; #B-H=2
     0  0  0  0 -1  0  0  1; #S-D=9
     0 -1  0  0  0  0  1  0; #H-R=7
    -1  0  0  0  0  0  0  0; #T-V=5
     0  0  0  1  0  0 -1  0; #I-H=5
     0  0  1  0 -1  0  0  0; #C-D=2
     1  0  0  0 -1  0  0  0; #V-D=2
     0 -1  0  0  0  0  0  1; #S-R=11
     0  0  0  1  0 -1  0  0; #I-B=5
     0  0  0  0  0 -1  0  0; #T-B=3
     0  0 -1  1  0  0  0  0] #I-C=7

b = [11,3,7,2,9,7,5,5,2,2,11,5,3,7]
# (AᵀA)x = Aᵀb eqs normais
x =(A'*A) \(A'*b)
println(x)
```
